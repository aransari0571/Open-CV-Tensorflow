```python
import tensorflow as tf 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array

datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale = None,
    fill_mode='reflect' # Changed from 'wrap' to 'nearest','reflect','constant'
)
```

```python
img = load_img(r"C:\Users\AR ANSARI\Downloads\panda image.jpg")  # PIL image
```

```python
img
```

```python
#convert to numpy array
x = img_to_array(img)  # shape (height, width, channels)
x = x.reshape((1,) + x.shape)  # shape (1, height, width, channels)
 
 # generate batches of augmented images from this image
i = 0
for batch in datagen.flow(x, batch_size=1,
                        save_to_dir=r"C:\Users\AR ANSARI\NIT\Deep Learning\CNN\OpenCV\data argumentation output"
                            save_prefix='panda', save_format='jpg'):
    i += 1
    if i > 20:
        break  # otherwise the generator would loop indefinitely
```
