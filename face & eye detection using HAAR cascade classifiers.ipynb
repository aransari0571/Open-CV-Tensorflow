# Face Detection


```python
import numpy as np
import cv2

# Load Haarcascade safely
face_classifier = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# Load image
image = cv2.imread(r"C:\Users\AR ANSARI\NIT\OpenCV\Haar Cascade\E IMG.jpg")

if image is None:
    print("Error: Image not found")
    exit()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

faces = face_classifier.detectMultiScale(gray, 1.3, 5)

if len(faces) == 0:
    print("No faces found")

else:
    for (x, y, w, h) in faces:
        cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 255), 2)

    cv2.imshow('Face Detected', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

# Eye Detection

```python

import numpy as np
import cv2

# Load Haarcascade safely
eye_classifier = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_eye.xml"
)

# Load image
image = cv2.imread(r"C:\Users\AR ANSARI\NIT\OpenCV\Haar Cascade\E IMG.jpg")

if image is None:
    print("Error: Image not found")
    exit()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Detect eyes
eyes = eye_classifier.detectMultiScale(gray, 1.3, 5)

if len(eyes) == 0:
    print("No eyes found")

else:
    for (x, y, w, h) in eyes:
        cv2.rectangle(image, (x, y), (x + w, y + h), (255, 0, 0), 2)  # Blue box for eyes

    cv2.imshow('Eye Detection', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

# Both Eye and Face Detection


```python
import cv2

# Automatically detect OpenCV’s cascade folder
cascade_path = cv2.data.haarcascades

face_classifier = cv2.CascadeClassifier(cascade_path + "haarcascade_frontalface_default.xml")
eye_classifier  = cv2.CascadeClassifier(cascade_path + "haarcascade_eye.xml")

if face_classifier.empty():
    print("Error: Face cascade not loaded.")
    exit()

if eye_classifier.empty():
    print("Error: Eye cascade not loaded.")
    exit()

img = cv2.imread(r"C:\Users\AR ANSARI\NIT\OpenCV\Haar Cascade\E IMG.jpg")

if img is None:
    print("Error: Image not found.")
    exit()

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

faces = face_classifier.detectMultiScale(gray, 1.3, 5)

if len(faces) == 0:
    print("No Face Found")

for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (127, 0, 255), 2)

    roi_gray = gray[y:y+h, x:x+w]
    roi_color = img[y:y+h, x:x+w]

    eyes = eye_classifier.detectMultiScale(roi_gray)

    for (ex, ey, ew, eh) in eyes:
        cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (255, 255, 0), 2)

cv2.imshow("img", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import gradio as gr

# Load Haar cascades (face + eyes only)
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye_tree_eyeglasses.xml")  # more accurate

def detect_face_eye(image):
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Detect faces only
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=6,
        minSize=(80, 80)
    )

    output = image.copy()

    # Loop through faces and then detect eyes inside face ROI
    for (x, y, w, h) in faces:
        cv2.rectangle(output, (x, y), (x+w, y+h), (255, 0, 0), 2)  # face box

        roi_gray = gray[y:y+h, x:x+w]
        roi_color = output[y:y+h, x:x+w]

        # Detect eyes only inside face area
        eyes = eye_cascade.detectMultiScale(
            roi_gray,
            scaleFactor=1.1,
            minNeighbors=15,     # High accuracy to remove false detection (nose)
            minSize=(40, 40)     # ignore small features like nose/mouth
        )

        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex+ew, ey+eh), (0, 255, 0), 2)  # eyes box

    return output

# Gradio Interface
app = gr.Interface(
    fn=detect_face_eye,
    inputs=gr.Image(sources=["upload", "webcam"], type="numpy"),
    outputs=gr.Image(type="numpy"),
    title="Face & Eye Detection (Only)",
    description="Upload an image or use webcam to detect only face and eyes using Haar Cascade",
    live=True
)

app.launch()
```
